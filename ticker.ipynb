```python
import yfinance as yf

try:
    # 1. Initialize the ticker object
    ticker_symbol = "IONQ"
    ticker = yf.Ticker(ticker_symbol)
    
    # 2. Access the fast info dictionary properties 
    # yfinance handles all behind-the-scenes token exchanges for you
    current_price = ticker.fast_info['lastPrice']
    currency = ticker.fast_info['currency']
    
    print(f"Ticker: {ticker_symbol}")
    print(f"Current Price: {current_price:.2f} {currency}")

except Exception as e:
    print(f"An error occurred while fetching Yahoo data: {e}")
```
