```python
from mem0 import Memory
memory = Memory()
import getpass

# 1. Securely prompt for your API key inside the notebook cell
# (It will show a hidden password input box)
openai_key = getpass.getpass("{{API_KEY}}")

# 2. Add the key directly into the Mem0 configuration dictionary
config = {
    "llm": {
        "provider": "openai",
        "config": {
            "model": "gpt-4o-mini",
            "api_key": openai_key  # <-- Your key is passed right here
        }
    },
    "embedder": {
        "provider": "openai",
        "config": {
            "model": "text-embedding-3-small",
            "api_key": openai_key  # <-- Passed here too for data vectorization
        }
    }
}

# 3. Initialize the memory layer with your localized configuration
memory = Memory.from_config(config)
```

```python
# 4. Run the rest of the script safely
raw_session_log = """
Doctor: Let's look at the patient's test results. FEV1 is down to 68%. 
Doctor: When we plot these multi-row medical subplots in Seaborn later, 
remember to always add plt.tight_layout() otherwise the axis text overlaps.
"""

print("\nProcessing and distilling memory...")
memory.add(raw_session_log, user_id="doctor_123")

print("\n--- SEARCHING DISTILLED MEDICAL KNOWLEDGE ---")
results = memory.search("What did we learn about lung metrics?", user_id="doctor_123")
for item in results:
    print(f"- {item['text']}")
```
